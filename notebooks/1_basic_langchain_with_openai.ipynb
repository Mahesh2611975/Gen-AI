## Load Env Variable

```python
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

### Conntect with Google Gemini model

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.7
)

response = llm.invoke("write a poem about the beauty of nature")
print(response.content)
```

### Conntect with OpenAi model

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-3.5-turbo",
    temperature=0.7
)

response = llm.invoke("write a poem about the beauty of nature")
print(response.content)
```

### Static prompts

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-3.5-turbo",
    temperature=0.7
)
```

```python
prompts =[
    ("system", "You are a python developer.") , ### "System","ai","User"
    ("user", "Write a python function to calculate the factorial of a number.")
]

response = llm.invoke(prompts)
print(response.content)
```
