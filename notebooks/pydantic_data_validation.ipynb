```python
from pydantic import BaseModel,Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name="mahesh", age =22)
user
```

### Field Validation

```python
class Product(BaseModel):
    title : str = Field(max_length=10)
    des : str = Field(min_length=20)
    price : int = Field(gt=0, le=100)
```

```python
product = Product(title="samsung", des="this is a amazing mobile from samsung...", price=90)
product2 = Product(title="mobile", des="this is a amazing mobile from samsung...", price=90)
```

```python
class Order(BaseModel):
    id:str = Field(min_length=6, max_length=10)
    product:Product
```

```python
order = Order(id="dmfdmf", product=product)
```

```python
order
```

```python
order.product.title
```

```python
class Carts(BaseModel):
    id: str = Field(min_length=6, max_length=10)
    items: List[Product]
    comment: str = Field(
        min_length=5,
        max_length=100
    )
```

```python
cart = Carts(id ="mdfjgnjg", items=[product,product2],comment="My shopping cart")
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

```python
### Annotated

commonValidator = Annotated[List[Product], Field(default=[])]
```

```python
class Carts(BaseModel):
    id: str = Field(min_length=6, max_length=10)
    items: commonValidator
    comment: str = Field(
        min_length=5,
        max_length=100
    )
```

```python
class Order(BaseModel):
    products:commonValidator
```

```python
ord = Order(products=[product,product2])
```
