```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"
]
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embaddings = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
vector = embaddings.embed_documents(documents)
```

```python
len(vector[0])
```

#### Store Vector Embaddings in Vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents,
    embedding=embaddings,
    persist_directory="./vector_db"
    

)
```

```python
query = "Machine learnings"
result = vector_store.similarity_search(query=query)
result
```
