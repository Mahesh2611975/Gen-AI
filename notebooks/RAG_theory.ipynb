## 1.RAG

Simple definition: RAG is a technique that **retrieves relevant information** from an external knowledge source and provides it to an AI model to generate a more accurate answer.

**RAG = Retrieval + Augmented + Generation**
- **Retrieval:** Finds relevant information from your data.
- **Augmented:** Adds that information to the AI's prompt as context.
- **Generation:** The AI generates an answer using the provided context.

# RAG Pipline

```text
             ┌──────────────────────┐
             │   Document Loading   │
             │     (PDF / TXT)      │ -->Load PDF, TXT, DOCX, or other data.
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │    Text Splitting    │
             │      (Chunks)        │ --> Divide large documents into smaller chunks
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │ Embedding Generation │ --> Convert each chunk into a numerical vector that represents its meaning. 
             │   (Vector Creation)  │
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │    Vector Database   │ --> Store the vectors and their associated text for similarity searching.
             │    (FAISS / Qdrant)  │
             └──────────┬───────────┘
                        │
                        │
             ┌──────────▼───────────┐
             │     User Query       │
             │  (Ask a Question)    │ --> The user enters a question.
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │      Retrieval       │ --> Search the database to find text relevant to the question.
             │ Relevant Text Chunks │
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │  Prompt Augmentation │
             │ Context + Question   │ --> Combine the retrieved text and the user's question into a prompt.
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │   LLM Generation     │
             │  (Answer Creation)   │ --The LLM uses the prompt to generate a response.
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │     Final Answer     │ --> Return the generated response to the user.
             └──────────────────────┘
```
