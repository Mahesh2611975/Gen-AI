```python
from langchain_openai import ChatOpenAI
from langchain_ollama import ChatOllama
llm = ChatOllama(
    model="qwen2.5:1.5b",
)
```

```python
prompts = [
    {"role": "system", "content": "You are a helpful assistant."},  
    {"role": "user", "content": "What is the capital of France?"}
]
res = llm.invoke(prompts)
print(res.content)
```

```python
prompts = [
    {"role": "system", "content": "You are a translator and translate input in hindi."},  
    {"role": "user", "content": "when you come to mumbai ?"}
]
res = llm.invoke(prompts)
print(res.content)
```

### Prompt Template - Dynamic Prompts

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm = ChatOpenAI(model="qwen2.5:1.5b")
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator and translate input in {language}."},
    {"role": "user", "content": "{query}"},
])
res = llm.invoke(prompts.format_messages(language="hindi", query="when you come to mumbai ?"))
print(res.content)
```
