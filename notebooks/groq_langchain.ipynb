## 1. Groq

Groq is an AI inference platform that provides very **fast access** to Large Language Models (LLMs).

-->Groq runs the model through its **cloud API.**

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
```

```python
llm =ChatGroq(model="openai/gpt-oss-20b")
```

```python
res = llm.invoke("can you tell me about the history of AI?")
```

```python
print(res.content)
```

```python
response = llm.invoke("explain the history of india?")
```

```python
print(response.content)
```

```python
response = llm.invoke("can you tell me about IPL?")
```

```python
print(response.content)
```

```python
response = llm.invoke("all winners of IPL since 2008?")
```

```python
print(response.content)
```
