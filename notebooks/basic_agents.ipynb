### 1.AI Agent

An AI Agent is an AI system that can understand a goal, decide what actions to take, use tools, and complete the task with limited human intervention.

AI Agent = LLM + Reasoning + Tools + Actions + Memory

#### 1.1 Main Components of an AI Agent

| Component | Simple Meaning | Example | Remember As |
|---|---|---|---|
|  **LLM** | Brain of the agent; understands and reasons | GPT, Gemini, Llama | **Think** |
|  **Tools** | Helps the agent interact with the outside world | Web Search, Calculator, API, SQL | **Use** |
|  **Reasoning** | Decides what action should be taken | "I need to search the database" | **Decide** |
|  **Memory** | Stores/recalls useful information | Chat history, Vector DB | **Remember** |
|  **Actions** | Performs the actual task | Send email, update DB, call API | **Do** |

#### AI Agent Workflow



```text
┌──────────────┐
│     USER     │
└──────┬───────┘
       ↓
┌──────────────┐
│    AGENT     │
└──────┬───────┘
       ↓
 Understand Goal
       ↓
  Decide Action
       ↓
┌──────────────┐
│     TOOL     │
└──────┬───────┘
       ↓
   Get Result
       ↓
 Analyze Result
       ↓
Need More Action?
   ↙          ↘
 YES          NO
  ↓            ↓

  ↑
  └──── Repeat ────┘
```

````python
from dotenv import load_dotenv
load_dotenv()
````

````python
from langchain_core.tools import tool
````

````python
@tool
def add_numbers(a: int, b: int):
    """
    it will return the sum of two numbers.

    Args:
        a : Number one
        b : Number two
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int):
    """
    it will return the product of two numbers.

    Args:
        a : Number one
        b : Number two
    """
    return a + b
````

````python
add_numbers.invoke({"a": 5, "b": 10})
````

````python
from langchain_groq import ChatGroq
llm =ChatGroq(model="openai/gpt-oss-20b", temperature=0, streaming=True)
````

````python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
````

````python
# llm = ChatOpenAI(model="gpt-4")
````

````python
agent =create_agent(
    model=llm,
    tools=[add_numbers,multiply_numbers],
    system_prompt ="you are a math teacyher, and always use tool for calculation."

)
````

````python
response = agent.invoke({"messages":[{"role":"user", "content":"what is 2 + 3 * 30?"}]})
print(response["messages"][-1].content)
````
