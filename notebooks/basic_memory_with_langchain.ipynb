```python
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI
# llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="qwen2.5:1.5b",
    temperature=0
)
```

```python
prompts =[
    {"role": "user", "content": "Hello, My name is Mahesh."},
    {"role": "ai", "content": "Hello Mahesh, nice to meet you! How can I assist you today?"},
    {"role": "user", "content":"What is My name"}

]
res = llm.invoke(prompts)
```

### Manage History

```python
history = []  ## [{"role": "user", "content": "Who is pm of India ?"}]
while True:
    query = input("User:")
    if query.lower() in ["exit", "quit", "byr"]:
        print("Good Bye")
        break



    history.append({"role":"user", "content":query})
    print("User :", query)

    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content})
    print("AI", res.content, "\n")
```
