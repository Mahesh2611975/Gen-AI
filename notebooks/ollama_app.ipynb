### Ollama

Ollama allows you to run models locally on your computer.

Your Computer

     ↓

   Ollama

     ↓

 Llama model

     ↓
     
 Response

generally we  don't need an internet connection after the model is downloaded.

```python
from langchain_ollama import ChatOllama

```

```python
llm = ChatOllama(model="qwen2.5:1.5b", temperature=0.7, max_tokens=2000)
```

```python
res = llm.invoke("What is the capitals of India and USA?")
res
```
