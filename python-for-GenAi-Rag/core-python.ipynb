## 1.7 Dictionaries

A **dictionary** stores data as **key-value** pairs.This is the shape of almost every API request and response you will touch.

--> Good for structured data

-->Dictionary → { }

-->Nested dictionary → { } inside a list/dictionary

```python
student = {
    "name": "Mahesh",
    "age": 21,
    "course": "CSE",
}

print(student["name"])       # direct access (KeyError if missing)
print(student.get("email"))  # safe access (None if missing)
```

```python
student = {"name": "Mahesh", "age": 21, "course": "CSE"}

print("keys       ->", student.keys())       # 1. all keys
print("values     ->", student.values())     # 2. all values
print("items      ->", student.items())      # 3. key-value pairs
print("get        ->", student.get("name"))  # 4. safe lookup

student.update({"age": 22})                  # 5. add / update
print("update     ->", student)

student.pop("age")                           # 6. remove a key
print("pop        ->", student)

student.popitem()                            # 7. remove the last pair
print("popitem    ->", student)

student.setdefault("city", "Kalaburagi")     # 8. get, or insert if missing
print("setdefault ->", student)

copy_of = student.copy()                     # 9. shallow copy
print("copy       ->", copy_of)

student.clear()                              # 10. empty the dict
print("clear      ->", student)
```

### A method that returns a value instead of printing

```python
class Chatbot:
    def __init__(self, name, temperature=0.7):
        self.name = name
        self.temperature = temperature

    def ask(self, question: str) -> str:
        return f"[{self.name} @ temp={self.temperature}] answering: {question}"

bot = Chatbot("GPT-5", temperature=0.2)
print(bot.ask("What is recursion?"))
```

### 2.7 Inheritance

**Inheritance** lets one class **reuse and extend** another class.

Inheritance means one class can **reuse the properties and methods** of another class.

```python
class Chatbot:

    def ask(self, question):
        print("Answering:", question)


class VoiceChatbot(Chatbot):

    def speak(self):
        print("Speaking...")


bot = VoiceChatbot()

bot.ask("What is AI?")
bot.speak()
```

### Overriding and super()

```python
class Chatbot:
    def __init__(self, name):
        self.name = name

    def ask(self, question):
        return f"{self.name}: {question}"


class VoiceChatbot(Chatbot):
    def __init__(self, name, voice):
        super().__init__(name)     # run the parent's __init__
        self.voice = voice

    def ask(self, question):       # override the parent's version
        base = super().ask(question)
        return f"{base} (spoken in {self.voice})"


bot = VoiceChatbot("GPT-5", voice="calm")
print(bot.ask("What is AI?"))
```

### 2.8 Encapsulation

Encapsulation means keeping data and the methods that work with that data together inside a class, while controlling how that data is accessed or changed.

Encapsulation keeps data and related functionality together and provides controlled access to that data.

```python
class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def show_balance(self):
        print("Balance:", self.balance)


account = BankAccount(1000)

account.deposit(500)
account.show_balance()
```

### 2.9 Polymorphism

Polymorphism means the **same method name** can behave differently depending on the object.

```python
class Chatbot:

    def respond(self):
        print("Text response")


class VoiceChatbot(Chatbot):

    def respond(self):
        print("Voice response")


class ImageChatbot(Chatbot):

    def respond(self):
        print("Image response")
text_bot = Chatbot()
voice_bot = VoiceChatbot()
image_bot = ImageChatbot()

text_bot.respond()
voice_bot.respond()
image_bot.respond()
```

### 2.10 Abstraction

Abstraction means **hiding unnecessary implementation details** and showing only **what the user needs to use.**

```python
# Python provides ABC and abstractmethod for creating abstract classes.
from abc import ABC, abstractmethod


class Chatbot(ABC):

    @abstractmethod
    def respond(self):
        pass
```

## 3. Async / Await

### 3.1 Why Asynchronous Programming Is Needed

**Asynchronous programming** lets a program start a long-running task (such as an API call) and keep doing other work while waiting for it to finish.

This matters for LLM applications because the bottleneck is almost always network latency, not CPU.



Your Program
    ↓
Send Request to LLM
    ↓
WAITING... ⏳
    ↓
Network Request / API Processing
    ↓
LLM Generates Response
    ↓
Receive Response
    ↓
Continue Your Program

### 3.2 Synchronous vs Asynchronous

**Synchronous:** tasks run one after another. The next task waits for the current one to finish.

Task 1 → wait → finish
                  ↓
              Task 2 → wait → finish
                              ↓
                          Task 3

```python
import time

def task(name):
    print("Starting", name)
    time.sleep(2)              # blocks the whole program
    print("Finished", name)

start = time.time()
task("A")
task("B")
print(f"Total: {time.time() - start:.1f}s")   # ~4 seconds
```

### 3.3 Asynchronous Execution

With async, tasks can wait **without blocking** the rest of the program.

Instead of sequential execution:

Task A -> wait -> finish -> Task B -> wait -> finish
asynchronous execution overlaps the waiting:

Task A -> waiting ─────────> finish
Task B -> waiting ─────────> finish
This is exactly what you want when firing several independent API calls.

### 3.4 async def

async def defines an **asynchronous function**, also called a coroutine function.

Calling it does **not** run it — it returns a coroutine object that must be awaited or run by an event loop.

```python
# normal function: calling it runs it
def hello():
    print("Hello")

hello()
```

```python
# async function: calling it returns a coroutine, nothing is printed yet
async def hello():
    print("Hello")

coro = hello()
print(coro)

coro.close()   # close it so Python does not warn about a never-awaited coroutine
```

The **async** keyword tells Python: this function can perform asynchronous operations.

### 3.5 await

await tells Python to wait for an asynchronous operation to finish **without blocking other async tasks.**

await asyncio.sleep(2) pauses only this coroutine for 2 seconds and lets other coroutines run. time.sleep(2) blocks the entire thread — very different.

```python
import asyncio

async def task():
    print("Starting")
    await asyncio.sleep(2)   # non-blocking pause
    print("Finished")
```

### 3.6 asyncio

asyncio is Python's built-in library for writing and running asynchronous programs. The three pieces used most often:

| **Function** | **Purpose** |
|---|---|
| `asyncio.run()` | Run a coroutine from normal (sync) code |
| `asyncio.sleep()` | Non-blocking pause |
| `asyncio.gather()` | Run several coroutines concurrently |

### 3.7 Running an Async Function

async def
   ↓
creates an async function (coroutine)
   ↓
asyncio.run()

⚠️ Inside Jupyter: an event loop is already running, so asyncio.run() raises RuntimeError: asyncio.run() cannot be called from a running event loop. In a notebook, just use await ... at the top level of a cell. In a normal .py script, use asyncio.run(main()).

### 3.9 Concurrent Execution

Concurrency means **handling multiple tasks** during overlapping periods instead of waiting for each task to completely finish before starting the next one.

Concurrency is what makes AI applications fast. Querying several LLMs, fetching embeddings, hitting a vector database — all of these are I/O bound.

Three independent calls, 2 seconds each:

Synchronous — 6 seconds

API 1: ██████████ (2s)
API 2:           ██████████ (2s)
API 3:                     ██████████ (2s)
-------------------------------------------
Total ≈ 6 seconds
Concurrent — 2 seconds

API 1: ██████████ (2s)
API 2: ██████████ (2s)
API 3: ██████████ (2s)
-------------------------------------------
Total ≈ 2 seconds
Key takeaway: concurrency overlaps the waiting, so total time ≈ the slowest single call.

### 3.10 asyncio.gather()

asyncio.gather() runs multiple coroutines concurrently and waits for all of them to finish, returning their results **in the order they were passed in.**

```python
import asyncio

async def task(name):
    print("Starting", name)
    await asyncio.sleep(2)
    print("Finished", name)
    return name
```

```python
import time

async def main():
    results = await asyncio.gather(
        task("A"),
        task("B"),
        task("C"),
    )
    print("Results:", results)

start = time.time()
await main()
print(f"Total: {time.time() - start:.1f}s")   # ~2s, not ~6s
```

### 3.11 Multiple LLM / API Calls Concurrently

When requests do not depend on each other, dispatch them all at once and collect the results when they are done.

             ┌──→ LLM 1 ──→ Response 1
             │
Question ────┼──→ LLM 2 ──→ Response 2
             │
             └──→ LLM 3 ──→ Response 3
                         │
                      gather
                         │
                         ▼
                   All responses

```python
import asyncio, time

# Simulated LLM call - in a real app this would be an HTTP request
async def call_llm(question: str) -> str:
    await asyncio.sleep(2)
    return f"Answer to: {question}"
```

```python
# Sequential (blocking): total = t1 + t2 + t3
start = time.time()

response1 = await call_llm("Question 1")
response2 = await call_llm("Question 2")
response3 = await call_llm("Question 3")

print(response1, "|", response2, "|", response3)
print(f"Sequential total: {time.time() - start:.1f}s")
```

```python
# Concurrent (non-blocking): total ~= slowest single call
start = time.time()

responses = await asyncio.gather(
    call_llm("Question 1"),
    call_llm("Question 2"),
    call_llm("Question 3"),
)

print(responses)
print(f"Concurrent total: {time.time() - start:.1f}s")
```
