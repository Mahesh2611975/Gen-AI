## 1. Core Python

*1.1 What is a variable?*

A variable is a name used to store a value in memory.

Think of a variable like a box. The box has a name, and inside it we store some data.

Python is dynamically typed — you don't declare the type, it is inferred from the value you assign.

```python
name = "Mahesh"
age = 21
course = "Python for GenAI"

print(type(name))
print(type(age))
print(type(course))
```

```python
# My personal details

name = "Mahesh"
age = 21
semester = 7
branch = "Computer Science and Engineering"
goal = "GenAI Developer"

print("Name:", name)
print("Age:", age)
print("Semester:", semester)
print("Branch:", branch)
print("Career Goal:", goal)
```
