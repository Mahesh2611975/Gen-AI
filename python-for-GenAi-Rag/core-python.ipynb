## 1.2 What is a data type?



| Data Type | Example | Used For |
|---|---|---|
| `str` | `"Mahesh"` | Text, names, messages, questions |
| `int` | `21` | Whole numbers, age, quantity, count |
| `float` | `5.8` | Decimal numbers, price, height, percentage |
| `bool` | `True` / `False` | Yes/no conditions and status |
| `list` | `["Python", "SQL", "RAG"]` | Multiple ordered values |
| `tuple` | `("Pune", "India")` | Multiple values that should not change |
| `dict` | `{"name": "Mahesh", "age": 21}` | Key-value data, JSON, API responses |
| `set` | `{"Python", "SQL"}` | Unique values and removing duplicates |
| `NoneType` | `None` | No value or empty result |

```python
name = "Mahesh"                         # str
age = 21                                # int
height = 5.8                            # float
is_student = True                       # bool
skills = ["Python"]                     # list
location = ("Pune", "India")            # tuple
student = {"name": "Mahesh"}            # dict
unique_numbers = {1, 2, 3}              # set
result = None                           # NoneType

for v in (name,age, height, is_student, skills, location,student,
          unique_numbers , result):
    print(f"{str(v):<20} -> {type(v).__name__}")
```

**The three you will use constantly:** str (prompts and responses), list (collections of messages or documents), dict (API payloads).

**1.3.1 if statement**

The if block runs only when the condition is True.

```python
age = 21

if age >= 18:
    print("You are eligible to vote")
```

```python
age = 15

if age <= 18:
    print("You are not eligible to vote")
```

**1.3.2 else statement**

else runs when the if condition is false.

```python
age = 15

if age >= 18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

**1.3.4. elif statement**

**elif means else if.** It allows us to check multiple conditions.

```python
marks = 75

if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Needs improvement")
```


### Combining conditions with and / or / not(Logical operator)

**and**

The and operator returns True only when both conditions are true.

```python
age = 21
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
else:
    print("Entry not allowed")
```

**or**

The or operator returns True when at least one condition is true.

```python
is_weekend = False
is_holiday = True

if is_weekend or is_holiday:
    print("No college today")
else:
    print("College today")
```

**not**

The not operator reverses a Boolean value.

```python
is_raining = False

if  not is_raining:
    print("You can go outside")
else:
    print("Take an umbrella")
```

## 1.4 Loops

A loop repeats a block of code.Loops are used when we want to execute the same code multiple times.

**.for** — iterate over a collection (most common)

**.while** — keep going while a condition stays true

### 1.4.1 for loop

for loop: A for loop is used to iterate over a sequence, such as a list, string, or range of numbers.

```python


for number in range(1, 6):
    print(number)
```

```python
# Loop through a list
skills = ["Python", "SQL", "FastAPI", "RAG"]

for skill in skills:
    print(skill)
```

```python
# Loop through a string
name = "Python"

for letter in name:
    print(letter)
```

```python
# range() Function :generates a sequence of numbers.
for number in range(5):
    print(number)
```

### 1.4.2 while Loop

A while loop runs as long as its condition is True.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
# break : break stops the loop immediately.
for number in range(1, 10):
    if number == 5:
        break
    print(number)

```

```python
# continue : continue skips the current iteration and moves to the next iteration.
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

```python
# enumerate() Function : it gives both 
# index
# value
skills = ["Python", "SQL", "FastAPI"]

for index, skill in enumerate(skills):
    print(index, skill)
```

```python
questions = [
    "What is Python?",
    "What is an API?",
    "What is RAG?"
]
for index, question in enumerate(questions, start=1):
    print(f"Question {index}: {question}")
```

## 1.6 Lists

--> A list is used to store multiple values in a single variable.

--> A list is an ordered, mutable collection of values.

--> the list are written inside the **Square bracket []**

```python
languages = ["Python", "Java", "C++"]

# indexing and slicing
print(languages[0])     # first
print(languages[-1])    # last
print(languages[0:2])   # first two
print(len(languages))   # length
```

```python
# Changing a List Element
names = ["Mahesh", "Sejal", "Rahul"]

names[2] = "Akash"

print(names)
```

```python
languages = ["Python", "Java", "C++"]

languages.append("JavaScript")        # 1. add an element at the end
print("append   ->", languages)

languages.insert(1, "C")              # 2. add at an index
print("insert   ->", languages)

languages.extend(["Go", "Rust"])      # 3. add many
print("extend   ->", languages)

languages.remove("Java")              # 4. remove by value
print("remove   ->", languages)

languages.pop(2)                      # 5. remove an elemetn by its index
print("pop      ->", languages)

print("index    ->", languages.index("Python"))   # 6. position of a value
print("count    ->", languages.count("Python"))   # 7. occurrences

languages.sort()                      # 8. sort in place
print("sort     ->", languages)

languages.reverse()                   # 9. reverse in place
print("reverse  ->", languages)

copy_of = languages.copy()            # 10. shallow copy
print("copy     ->", copy_of)

languages.clear()                     # 11. empty the list
print("clear    ->", languages)
```

```python
documents = ["doc1", "doc2", "doc3"]

upper = [d.upper() for d in documents]
print(upper)
```

## 1.7 Dictionaries

A **dictionary** stores data as **key-value** pairs.This is the shape of almost every API request and response you will touch.

--> Good for structured data

-->Dictionary → { }

-->Nested dictionary → { } inside a list/dictionary

```python
student = {
    "name": "Mahesh",
    "age": 21,
    "course": "CSE",
}

print(student["name"])       # direct access (KeyError if missing)
print(student.get("email"))  # safe access (None if missing)
```

```python
student = {"name": "Mahesh", "age": 21, "course": "CSE"}

print("keys       ->", student.keys())       # 1. all keys
print("values     ->", student.values())     # 2. all values
print("items      ->", student.items())      # 3. key-value pairs
print("get        ->", student.get("name"))  # 4. safe lookup

student.update({"age": 22})                  # 5. add / update
print("update     ->", student)

student.pop("age")                           # 6. remove a key
print("pop        ->", student)

student.popitem()                            # 7. remove the last pair
print("popitem    ->", student)

student.setdefault("city", "Kalaburagi")     # 8. get, or insert if missing
print("setdefault ->", student)

copy_of = student.copy()                     # 9. shallow copy
print("copy       ->", copy_of)

student.clear()                              # 10. empty the dict
print("clear      ->", student)
```

### Nested Dictionary

A dictionary can contain another dictionary.

```python
student = {
    "name": "Mahesh",
    "details": {
        "age": 21,
        "course": "CSE"
    }
}
print(student["details"]["course"])
```

### Looping over a dictionary

```python
config = {"model": "gpt-5", "temperature": 0.7, "max_tokens": 500}

for key, value in config.items():
    
    print(f"{key:<12} = {value}")
```

## 1.8 File Handling

File handling means reading data from files and writing data into files using Python.

The standard pattern uses **with**, which closes the file automatically even if an error happens.



| Mode | Meaning |
|---|---|
| `"r"` | Read |
| `"w"` | Write / overwrite |
| `"a"` | Append |
| `"r+"` | Read + write |

```python
# Reading the whole file
with open("notes.txt", "r") as file:
    content = file.read()

print(content)
```

```python
## Writing to a File
with open("notes.txt", "w") as file:
    file.write("Python is useful for GenAI.")
    file.write("Second line\n")
```

```python
# Append to a File --> add content without removing existing content
with open("notes.txt", "a") as file:
    file.write("\nI am learning RAG.")
```

## 1.9 JSON

JSON (JavaScript Object Notation)is a text format for exchanging structured data. Every LLM API speaks JSON.

```python
{
    "name": "Mahesh",
    "age": 21,
    "skills": ["Python", "SQL", "FastAPI"]
}
```

| **Function** | **Direction** |
|---|---|
| `json.loads()` | JSON string → Python object |
| `json.dumps()` | Python object → JSON string |
| `json.load()` | JSON file → Python object |
| `json.dump()` | Python object → JSON file |

```python
import json

# json.loads() : JSON string -> Python dict
data = '{"name": "Mahesh", "age": 22}'
user = json.loads(data)

print(user["name"])
print(type(user))
```

```python
import json

# json.dumps() : Python dict -> JSON string
user = {"name": "Nitish", "age": 22, "skills": ["Python", "Java"]}
data = json.dumps(user, indent=2)

print(data)
print(type(data))
```
