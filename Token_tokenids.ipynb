```python
text = "who is pm of India ?"
```

```python
import tiktoken
```

```python
tokenizer = tiktoken.encoding_for_model(model_name="gpt-4")
```

```python
tokenids = tokenizer.encode(text)
tokenids
```

```python
resultIds = [47, 76, 315, 6890, 374, 5768, 35973]
```

```python
result = tokenizer.decode(resultIds)
result
```
